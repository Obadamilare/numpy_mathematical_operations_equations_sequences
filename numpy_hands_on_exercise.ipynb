## Exercise 2: Student Performance Analysis

### Task 1: Create and Inspect the Data

```python
scores = np.array([62, 75, 81, 69, 88, 94, 73, 85, 77, 91])
print(scores)
print(type(scores))
print(scores.shape)
```

### Task 2: Calculate the Mean Score

```python
mean_score = np.mean(scores)
print(mean_score)
```

### Task 3: Calculate Each Student's Deviation

```python
deviations = scores - mean_score
print(deviations)
```

### Task 4: Identify Positive and Negative Deviations

A positive deviation means that student scored above the class average.
A negative deviation means that student scored below the class average.
A value close to zero means that student scored close to the average, a typical result for this class.

### Task 5: Square the Deviations

```python
squared_deviations = deviations ** 2
print(squared_deviations)
```

Why squaring removes the negative signs: a negative number times a negative number gives a positive number. So squaring every deviation makes all values positive. This matters because the plain deviations always add up to zero by definition of the mean. Squaring stops the positive and negative values from cancelling out, so we can measure the spread.

### Task 6: Calculate the Sum of Squared Deviations

```python
sum_squared_deviations = np.sum(squared_deviations)
print(sum_squared_deviations)
```

### Task 7: Connect the Steps to Standard Deviation

The steps to build standard deviation are:

1. Calculate the mean of the data.
2. Calculate each value's deviation from the mean (value minus mean).
3. Square each deviation, so every value becomes positive.
4. Sum the squared deviations.
5. Divide by the number of values, n. This gives the variance.
6. Take the square root of the variance. This gives the standard deviation.

```python
variance = sum_squared_deviations / len(scores)
standard_deviation = np.sqrt(variance)
print(variance)
print(standard_deviation)

print(np.std(scores))
```

Why standard deviation matters here: the mean alone cannot tell us if every student scored close to it, or if the class is a mix of very strong and very weak students that happen to average out to the same number. Standard deviation measures that spread. A small value means most students scored close together. A large value means there is a wide gap between the top and the struggling students, which usually needs a different teaching response.

Industry Thinking Challenge

The average alone hides how individual students performed. Two classes can share the same average score but have very different needs. One class might be evenly prepared, while the other might have a few top scorers pulling up a mostly struggling group. Looking at individual deviations, and the standard deviation as a summary, helps the coordinator see which students need extra support and whether the class is learning consistently.

## Exercise 3: Trigonometric Series and Convergence


### Task 1: Define the Angle

```python
theta = 30
theta_radians = np.radians(theta)
print(theta_radians)
```

### Task 2: Create the Sequence of Terms

```python
k = np.arange(1, 100, 1)
print(k)
print(k[0])
print(k[-1])
print(k.size)
```

### Task 3: Calculate the Trigonometric Component

```python
sin_theta = np.sin(theta_radians)
print(sin_theta)
```

### Task 4: Build the Series Terms

```python
terms = sin_theta / k
print(terms)
```

### Task 5: Calculate the Sum

```python
series_sum = np.sum(terms)
print(series_sum)
```

### Task 6: Increase the Number of Terms

```python
k_100 = np.arange(1, 100, 1)
k_1000 = np.arange(1, 1000, 1)
k_10000 = np.arange(1, 10000, 1)

sum_100 = np.sum(sin_theta / k_100)
sum_1000 = np.sum(sin_theta / k_1000)
sum_10000 = np.sum(sin_theta / k_10000)

print(sum_100)
print(sum_1000)
print(sum_10000)
```

### Task 8: Explain Convergence

1. With fewer terms (99), the sum was a small value, about 2.59.
2. With more terms (999, then 9,999), the sum did not settle near that same number. It kept getting bigger each time.
3. Why it keeps changing: every term, sin(theta) divided by k, is positive, because sin(30 degrees) is a fixed positive number, 0.5. Since every term we add is positive, the running total can only grow. It never levels off.
4. Does it reach a limiting value? No. This is the key point. This series does not converge. The sum of 1/k from k = 1 to n is called the harmonic series, and it is a well known result that the harmonic series diverges. It grows without any upper limit, just very slowly, roughly like the natural logarithm of n. Multiplying every term by the constant 0.5 does not change that. It only scales the growth by half.

Simple takeaway: a series can look like it is settling down just because it grows slowly. That is not the same as truly converging. A series only converges if, as more terms are added, the sum gets closer and closer to one fixed number and stays there. Here, no matter how many terms we add, the sum will eventually pass any number we name. It just takes a huge number of terms to get very high. For comparison, a series like sin(theta) divided by k squared, instead of k, would actually converge to a fixed value as terms increase.

## Bonus Challenge: Mini NumPy Numerical Experiment

Project title: Solar Panel Output Analysis, Weekly Energy Yield

Scenario: Energy. A small solar installation team wants to understand how consistent their panel's daily output has been over 10 days, and how a proposed cleaning upgrade, expected to boost output by 12%, would change things.

Objective: use NumPy to summarize the dataset, measure the spread of daily output with standard deviation, and estimate the effect of the upgrade.

Dataset: 10 days of recorded energy output, in kWh.

```python
daily_output = np.array([14.2, 13.8, 15.1, 12.9, 16.0, 15.5, 11.7, 14.9, 13.4, 15.8])
days = np.arange(1, 11, 1)
print(daily_output)
print(days)
```

Calculation 1: total and average output

```python
total_output = np.sum(daily_output)
mean_output = np.mean(daily_output)
print(total_output)
print(mean_output)
```

Calculation 2: spread of the output, standard deviation

```python
deviations = np.subtract(daily_output, mean_output)
squared_deviations = deviations ** 2
variance = np.mean(squared_deviations)
std_dev = np.sqrt(variance)
print(std_dev)
```

Calculation 3: projected output after a 12% upgrade

```python
projected_output = np.multiply(daily_output, 1.12)
gain = projected_output - daily_output
total_gain = np.sum(gain)
print(projected_output)
print(total_gain)
```

Results and Interpretation

Over the 10 day period, the panel produced about 143.3 kWh total, averaging about 14.33 kWh per day. The standard deviation is about 1.32 kWh, which means daily output is fairly steady, with no extreme outlier days. This is a good sign for reliability planning. If the cleaning upgrade delivers the estimated 12% gain, projected daily output rises across the board, adding about 17.2 kWh total over the same 10 days. This gives the team a clear number to weigh against the cost of the upgrade.

Conclusion: the low standard deviation shows this panel performs steadily from day to day. A 12% output increase from the cleaning upgrade would translate almost directly into a reliable 12% gain in usable energy, which makes it a low risk investment to look into further.
